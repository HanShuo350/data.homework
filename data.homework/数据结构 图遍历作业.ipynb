我来分析这个无向图，写出从顶点 **A** 出发的 **DFS** 和 **BFS** 遍历序列。

---

## 第一步：确定图中的边和邻接关系

根据图像，5个顶点（A, B, C, D, E）之间的边：

| 边 | 连接的顶点 |
|---|---|
| 1 | A — B |
| 2 | A — C |
| 3 | B — D |
| 4 | C — D |
| 5 | C — E |
| 6 | D — E |

**各顶点的邻接点（按字母顺序排列）：**

| 顶点 | 邻接点（字母顺序） |
|:---|:---|
| A | B, C |
| B | A, D |
| C | A, D, E |
| D | B, C, E |
| E | C, D |

---

## 第二步：DFS（深度优先搜索）

**规则**：从起点出发，优先访问**最深的未访问节点**（递归/栈），邻接点按字母顺序访问。

```
1. 访问 A，标记 A 已访问
   A 的邻接点：B, C → 先访问 B

2. 访问 B，标记 B 已访问
   B 的邻接点：A(已访问), D → 访问 D

3. 访问 D，标记 D 已访问
   D 的邻接点：B(已访问), C, E → 按字母顺序先访问 C

4. 访问 C，标记 C 已访问
   C 的邻接点：A(已访问), D(已访问), E → 访问 E

5. 访问 E，标记 E 已访问
   E 的邻接点：C(已访问), D(已访问) → 全部已访问，回溯

   回溯到 C → 全部已访问
   回溯到 D → 全部已访问
   回溯到 B → 全部已访问
   回溯到 A → 全部已访问
```

### DFS 序列：**A → B → D → C → E**

---

## 第三步：BFS（广度优先搜索）

**规则**：从起点出发，优先访问**同一层的所有节点**（队列），邻接点按字母顺序入队。

```
1. 访问 A，标记 A 已访问
   将 A 的邻接点按字母顺序入队：B, C
   队列：[B, C]

2. 出队 B，访问 B，标记 B 已访问
   B 的邻接点：A(已访问), D → D 入队
   队列：[C, D]

3. 出队 C，访问 C，标记 C 已访问
   C 的邻接点：A(已访问), D(已入队), E → E 入队
   队列：[D, E]

4. 出队 D，访问 D，标记 D 已访问
   D 的邻接点：B(已访问), C(已访问), E(已入队) → 无新节点入队
   队列：[E]

5. 出队 E，访问 E，标记 E 已访问
   E 的邻接点：C(已访问), D(已访问) → 无新节点入队
   队列：[]
```

### BFS 序列：**A → B → C → D → E**

---

## 最终答案

| 遍历方式 | 序列 |
|:---|:---|
| **DFS** | **A, B, D, C, E** |
| **BFS** | **A, B, C, D, E** |

````python
from collections import deque

# 定义图（邻接表表示）——邻接点按字母顺序排列
graph = {
    'A': ['B', 'C'],
    'B': ['A', 'D'],
    'C': ['A', 'D', 'E'],
    'D': ['B', 'C', 'E'],
    'E': ['C', 'D']
}

# DFS（递归实现）
def dfs(graph, start):
    visited = set()
    result = []

    def dfs_helper(vertex):
        visited.add(vertex)
        result.append(vertex)
        for neighbor in graph[vertex]:      # 邻接点已按字母顺序排列
            if neighbor not in visited:
                dfs_helper(neighbor)

    dfs_helper(start)
    return result

# DFS（迭代实现，使用栈）
def dfs_iterative(graph, start):
    visited = set()
    stack = [start]
    result = []

    while stack:
        vertex = stack.pop()
        if vertex not in visited:
            visited.add(vertex)
            result.append(vertex)
            # 逆序入栈，保证按字母顺序访问
            for neighbor in reversed(graph[vertex]):
                if neighbor not in visited:
                    stack.append(neighbor)

    return result

# BFS（使用队列）
def bfs(graph, start):
    visited = set()
    queue = deque([start])
    visited.add(start)
    result = []

    while queue:
        vertex = queue.popleft()
        result.append(vertex)
        for neighbor in graph[vertex]:      # 邻接点已按字母顺序入队
            if neighbor not in visited:
                visited.add(neighbor)
                queue.append(neighbor)

    return result

# 执行
start = 'A'
print("DFS 序列:", " → ".join(dfs(graph, start)))        # A → B → D → C → E
print("BFS 序列:", " → ".join(bfs(graph, start)))        # A → B → C → D → E
````
