       /  \
      5    15
     / \     \
    3   7     20

```python
from typing import List, Optional


class TreeNode:
    """二叉树节点类（链表结构）"""
    def __init__(self, val=0, left=None, right=None):
        self.val = val
        self.left = left
        self.right = right


def build_tree_from_level_order(arr: List[Optional[int]]) -> Optional[TreeNode]:
    """
    从层序遍历数组还原二叉树

    参数:
        arr: 层序遍历数组，None 表示空节点
        例如: [10, 5, 15, 3, 7, None, 20]

    返回:
        二叉树的根节点
    """
    if not arr or arr[0] is None:
        return None

    # 创建根节点
    root = TreeNode(arr[0])

    # 使用队列保存待处理父节点
    queue = [root]
    i = 1  # 数组索引，指向当前要处理的子节点

    while queue and i < len(arr):
        node = queue.pop(0)  # 取出队首节点

        # 处理左子节点
        if i < len(arr) and arr[i] is not None:
            node.left = TreeNode(arr[i])
            queue.append(node.left)
        i += 1

        # 处理右子节点
        if i < len(arr) and arr[i] is not None:
            node.right = TreeNode(arr[i])
            queue.append(node.right)
        i += 1

    return root


def print_tree_structure(root: Optional[TreeNode], prefix: str = "", is_left: bool = True):
    """
    打印树的结构（可视化形态）

    参数:
        root: 根节点
        prefix: 前缀（用于缩进）
        is_left: 是否为左子树
    """
    if root is None:
        return

    # 打印当前节点
    print(prefix + ("├── " if is_left else "└── ") + str(root.val))

    # 计算子节点的前缀
    child_prefix = prefix + ("│   " if is_left else "    ")

    # 递归打印左右子树
    if root.left or root.right:
        if root.left:
            print_tree_structure(root.left, child_prefix, True)
        else:
            # 打印空左子树占位符
            print(child_prefix + "├── None")

        if root.right:
            print_tree_structure(root.right, child_prefix, False)
        else:
            # 打印空右子树占位符
            print(child_prefix + "└── None")


def inorder_traversal(root: Optional[TreeNode]) -> List[int]:
    """中序遍历（左-根-右）"""
    result = []

    def inorder(node):
        if node:
            inorder(node.left)
            result.append(node.val)
            inorder(node.right)

    inorder(root)
    return result


def preorder_traversal(root: Optional[TreeNode]) -> List[int]:
    """前序遍历（根-左-右）"""
    result = []

    def preorder(node):
        if node:
            result.append(node.val)
            preorder(node.left)
            preorder(node.right)

    preorder(root)
    return result


def postorder_traversal(root: Optional[TreeNode]) -> List[int]:
    """后序遍历（左-右-根）"""
    result = []

    def postorder(node):
        if node:
            postorder(node.left)
            postorder(node.right)
            result.append(node.val)

    postorder(root)
    return result


def level_order_traversal(root: Optional[TreeNode]) -> List[Optional[int]]:
    """层序遍历（验证还原结果）"""
    if not root:
        return []

    result = []
    queue = [root]

    while queue:
        node = queue.pop(0)
        if node:
            result.append(node.val)
            queue.append(node.left)
            queue.append(node.right)
        else:
            result.append(None)

    # 移除末尾的连续 None（可选，便于查看）
    while result and result[-1] is None:
        result.pop()

    return result


# ==================== 主程序 ====================
if __name__ == "__main__":
    # 给定的层序遍历数组
    arr = [10, 5, 15, 3, 7, None, 20]

    print("=" * 60)
    print("给定的层序遍历数组:", arr)
    print("=" * 60)

    # 还原二叉树
    root = build_tree_from_level_order(arr)

    # 打印树的形态
    print("\n【树的形态】")
    print("      10")
    print("     /  \\")
    print("    5    15")
    print("   / \\     \\")
    print("  3   7     20")
    print("\n使用树形结构打印：")
    print_tree_structure(root)

    # 验证遍历结果
    print("\n【遍历验证】")
    print(f"前序遍历 (根-左-右): {preorder_traversal(root)}")
    print(f"中序遍历 (左-根-右): {inorder_traversal(root)}")
    print(f"后序遍历 (左-右-根): {postorder_traversal(root)}")
    print(f"层序遍历还原结果: {level_order_traversal(root)}")

    # 验证与原始数组一致
    reconstructed = level_order_traversal(root)
    print(f"\n【验证】还原后的层序遍历: {reconstructed}")

    # 手动检查节点关系
    print("\n【节点关系验证】")
    print(f"根节点: {root.val}")
    print(f"10的左孩子: {root.left.val if root.left else None}")
    print(f"10的右孩子: {root.right.val if root.right else None}")
    print(f"5的左孩子: {root.left.left.val if root.left and root.left.left else None}")
    print(f"5的右孩子: {root.left.right.val if root.left and root.left.right else None}")
    print(f"15的左孩子: {root.right.left.val if root.right and root.right.left else None}")
    print(f"15的右孩子: {root.right.right.val if root.right and root.right.right else None}")

    print("\n" + "=" * 60)
    print("还原完成！")
    print("=" * 60)
```
