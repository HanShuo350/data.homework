```python
class ListNode:
    """链表结点类"""
    def __init__(self, val=0, next=None):
        self.val = val
        self.next = next


class LinkedList:
    """链表类，包含反转和检测环的方法"""
    def __init__(self):
        self.head = None

    def append(self, val):
        """在链表末尾添加结点"""
        if not self.head:
            self.head = ListNode(val)
            return
        curr = self.head
        while curr.next:
            curr = curr.next
        curr.next = ListNode(val)

    def reverse(self):
        """
        反转链表（方法1：迭代法）
        时间复杂度：O(n)，空间复杂度：O(1)
        例如：[a, b, c] → [c, b, a]
        """
        prev = None
        curr = self.head

        while curr:
            next_temp = curr.next  # 保存下一个结点
            curr.next = prev       # 反转指针
            prev = curr            # 移动prev
            curr = next_temp       # 移动curr

        self.head = prev  # 更新头结点

    def reverse_recursive(self):
        """
        反转链表（方法2：递归法）
        时间复杂度：O(n)，空间复杂度：O(n)（递归栈）
        """
        def _reverse(node, prev=None):
            if not node:
                return prev
            next_node = node.next
            node.next = prev
            return _reverse(next_node, node)

        self.head = _reverse(self.head)

    def has_cycle(self):
        """
        检测链表中是否存在环（快慢指针法）
        时间复杂度：O(n)，空间复杂度：O(1)
        返回：True（有环） / False（无环）
        """
        if not self.head or not self.head.next:
            return False

        slow = self.head      # 慢指针，每次走一步
        fast = self.head      # 快指针，每次走两步

        while fast and fast.next:
            slow = slow.next
            fast = fast.next.next
            if slow == fast:   # 相遇说明有环
                return True

        return False

    def to_list(self):
        """将链表转换为Python列表（便于测试）"""
        result = []
        curr = self.head
        # 注意：为了安全，如果检测到环，最多遍历1000个结点
        count = 0
        while curr and count < 1000:
            result.append(curr.val)
            curr = curr.next
            count += 1
        return result

```

```python
# ==================== 测试代码 ====================
if __name__ == "__main__":
    print("=" * 50)
    print("测试1：链表反转")
    print("=" * 50)

    # 创建链表 [a, b, c] → [c, b, a]
    ll = LinkedList()
    for val in ['a', 'b', 'c']:
        ll.append(val)

    print(f"反转前：{ll.to_list()}")
    ll.reverse()
    print(f"反转后：{ll.to_list()}")

    # 测试递归反转
    print("\n测试递归反转：")
    ll2 = LinkedList()
    for val in [1, 2, 3, 4, 5]:
        ll2.append(val)
    print(f"反转前：{ll2.to_list()}")
    ll2.reverse_recursive()
    print(f"递归反转后：{ll2.to_list()}")

    print("\n" + "=" * 50)
    print("测试2：检测环")
    print("=" * 50)

    # 测试1：无环链表
    ll_no_cycle = LinkedList()
    for val in [1, 2, 3, 4, 5]:
        ll_no_cycle.append(val)
    print(f"链表 [1,2,3,4,5] 是否有环？{ll_no_cycle.has_cycle()}")

    # 测试2：有环链表（手动创建）
    node1 = ListNode(1)
    node2 = ListNode(2)
    node3 = ListNode(3)
    node4 = ListNode(4)
    node5 = ListNode(5)

    node1.next = node2
    node2.next = node3
    node3.next = node4
    node4.next = node5
    node5.next = node3  # 创建环：5指向3

    ll_with_cycle = LinkedList()
    ll_with_cycle.head = node1
    print(f"链表 [1,2,3,4,5] 且 5→3 是否有环？{ll_with_cycle.has_cycle()}")

    # 测试3：单结点无环
    ll_single = LinkedList()
    ll_single.append(42)
    print(f"单结点 [42] 是否有环？{ll_single.has_cycle()}")

    # 测试4：单结点自环
    single_cycle = ListNode(99)
    single_cycle.next = single_cycle  # 自环
    ll_self_cycle = LinkedList()
    ll_self_cycle.head = single_cycle
    print(f"单结点自环 [99→自己] 是否有环？{ll_self_cycle.has_cycle()}")

    print("\n" + "=" * 50)
    print("所有测试完成！")
    print("=" * 50)
```
