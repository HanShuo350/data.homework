```python
from typing import Optional, List


class TreeNode:
    """二叉搜索树节点类"""
    def __init__(self, val: int):
        self.val = val
        self.left: Optional[TreeNode] = None
        self.right: Optional[TreeNode] = None


class BST:
    """二叉搜索树"""
    def __init__(self):
        self.root: Optional[TreeNode] = None

    def insert(self, val: int) -> None:
        """插入节点"""
        if self.root is None:
            self.root = TreeNode(val)
            return

        curr = self.root
        while curr:
            if val < curr.val:
                if curr.left is None:
                    curr.left = TreeNode(val)
                    return
                curr = curr.left
            elif val > curr.val:
                if curr.right is None:
                    curr.right = TreeNode(val)
                    return
                curr = curr.right
            else:
                # 值已存在，不插入
                return

    def delete_root_with_predecessor(self) -> Optional[int]:
        """
        使用【中序前驱】策略删除根节点
        中序前驱：左子树中的最大值（最右节点）

        步骤：
        1. 如果根节点为空，返回 None
        2. 如果根节点没有左子树，直接用右子树作为新根
        3. 否则，找到左子树中的最大值节点（前驱）
        4. 用前驱的值替换根节点的值
        5. 删除前驱节点（递归删除）
        """
        if self.root is None:
            return None

        deleted_val = self.root.val

        # 情况1：没有左子树，直接用右子树作为新根
        if self.root.left is None:
            self.root = self.root.right
            return deleted_val

        # 情况2：没有右子树，直接用左子树作为新根
        if self.root.right is None:
            self.root = self.root.left
            return deleted_val

        # 情况3：左右子树都存在，使用中序前驱
        # 找到左子树中的最大值节点及其父节点
        parent = self.root
        predecessor = self.root.left

        # 走到左子树的最右边
        while predecessor.right:
            parent = predecessor
            predecessor = predecessor.right

        # 用前驱的值替换根节点的值
        self.root.val = predecessor.val

        # 删除前驱节点
        if parent.left == predecessor:
            parent.left = predecessor.left
        else:
            parent.right = predecessor.left

        return deleted_val

    def delete_root_with_successor(self) -> Optional[int]:
        """
        使用【中序后继】策略删除根节点
        中序后继：右子树中的最小值（最左节点）

        步骤：
        1. 如果根节点为空，返回 None
        2. 如果根节点没有右子树，直接用左子树作为新根
        3. 否则，找到右子树中的最小值节点（后继）
        4. 用后继的值替换根节点的值
        5. 删除后继节点（递归删除）
        """
        if self.root is None:
            return None

        deleted_val = self.root.val

        # 情况1：没有右子树，直接用左子树作为新根
        if self.root.right is None:
            self.root = self.root.left
            return deleted_val

        # 情况2：没有左子树，直接用右子树作为新根
        if self.root.left is None:
            self.root = self.root.right
            return deleted_val

        # 情况3：左右子树都存在，使用中序后继
        # 找到右子树中的最小值节点及其父节点
        parent = self.root
        successor = self.root.right

        # 走到右子树的最左边
        while successor.left:
            parent = successor
            successor = successor.left

        # 用后继的值替换根节点的值
        self.root.val = successor.val

        # 删除后继节点
        if parent.left == successor:
            parent.left = successor.right
        else:
            parent.right = successor.right

        return deleted_val

    def inorder_traversal(self) -> List[int]:
        """中序遍历（返回有序序列）"""
        result = []

        def inorder(node: Optional[TreeNode]):
            if node:
                inorder(node.left)
                result.append(node.val)
                inorder(node.right)

        inorder(self.root)
        return result

    def print_tree_structure(self, node: Optional[TreeNode] = None, prefix: str = "", is_left: bool = True, is_root: bool = True):
        """
        打印树的结构（可视化形态）
        """
        if node is None:
            node = self.root

        if node is None:
            print("空树")
            return

        if is_root:
            print(str(node.val))
            # 打印左子树
            if node.left or node.right:
                if node.left:
                    self.print_tree_structure(node.left, "", True, False)
                else:
                    print("├── None")

                if node.right:
                    self.print_tree_structure(node.right, "", False, False)
                else:
                    print("└── None")
        else:
            print(prefix + ("├── " if is_left else "└── ") + str(node.val))
            child_prefix = prefix + ("│   " if is_left else "    ")
            if node.left or node.right:
                if node.left:
                    self.print_tree_structure(node.left, child_prefix, True, False)
                else:
                    print(child_prefix + "├── None")
                if node.right:
                    self.print_tree_structure(node.right, child_prefix, False, False)
                else:
                    print(child_prefix + "└── None")

    def get_ascii_tree(self) -> str:
        """生成ASCII艺术风格的树形图"""
        lines = []

        def build_lines(node: Optional[TreeNode], prefix: str = "", is_left: bool = True):
            if node is None:
                return
            lines.append(prefix + ("├── " if is_left else "└── ") + str(node.val))
            child_prefix = prefix + ("│   " if is_left else "    ")
            build_lines(node.left, child_prefix, True)
            build_lines(node.right, child_prefix, False)

        if self.root:
            lines.append(str(self.root.val))
            child_prefix = ""
            if self.root.left:
                build_lines(self.root.left, child_prefix, True)
            if self.root.right:
                build_lines(self.root.right, child_prefix, False)
        return "\n".join(lines)


# ==================== 主程序 ====================
if __name__ == "__main__":
    print("=" * 70)
    print("题1：构建 BST")
    print("=" * 70)

    # 插入序列
    insert_sequence = [50, 30, 70, 20, 40, 60, 80]
    bst = BST()

    print(f"依次插入序列: {insert_sequence}")
    print("\n插入过程：")
    print("插入 50 → 根节点")
    print("插入 30 → 50的左孩子")
    print("插入 70 → 50的右孩子")
    print("插入 20 → 30的左孩子")
    print("插入 40 → 30的右孩子")
    print("插入 60 → 70的左孩子")
    print("插入 80 → 70的右孩子")

    for val in insert_sequence:
        bst.insert(val)

    print("\n" + "-" * 70)
    print("最终 BST 形态：")
    print("-" * 70)
    print(bst.get_ascii_tree())

    print("\n" + "-" * 70)
    print("树的层级结构表示：")
    print("-" * 70)
    print("        50")
    print("       /  \\")
    print("      30   70")
    print("     / \\   / \\")
    print("    20 40 60 80")

    print(f"\n中序遍历结果（验证BST性质）: {bst.inorder_traversal()}")

    print("\n" + "=" * 70)
    print("题2：删除根结点 50")
    print("=" * 70)

    # ===== 方法1：使用中序前驱 =====
    print("\n【方法1】使用【中序前驱】策略删除根节点")
    print("-" * 70)
    print("中序前驱定义：左子树中的最大值（最右节点）")
    print("左子树 [30, 20, 40] 中的最大值是 40")
    print("步骤：")
    print("1. 找到前驱节点 40")
    print("2. 用 40 替换根节点的值")
    print("3. 删除原来的前驱节点 40")

    bst1 = BST()
    for val in insert_sequence:
        bst1.insert(val)

    print("\n原 BST：")
    print(bst1.get_ascii_tree())

    deleted1 = bst1.delete_root_with_predecessor()
    print(f"\n删除的根节点值: {deleted1}")

    print("\n删除后 BST 形态：")
    print(bst1.get_ascii_tree())

    print("\n树的层级结构表示：")
    print("        40")
    print("       /  \\")
    print("      30   70")
    print("     /    / \\")
    print("    20   60 80")

    print(f"\n中序遍历结果: {bst1.inorder_traversal()}")

    # ===== 方法2：使用中序后继 =====
    print("\n" + "-" * 70)
    print("【方法2】使用【中序后继】策略删除根节点")
    print("-" * 70)
    print("中序后继定义：右子树中的最小值（最左节点）")
    print("右子树 [70, 60, 80] 中的最小值是 60")
    print("步骤：")
    print("1. 找到后继节点 60")
    print("2. 用 60 替换根节点的值")
    print("3. 删除原来的后继节点 60")

    bst2 = BST()
    for val in insert_sequence:
        bst2.insert(val)

    print("\n原 BST：")
    print(bst2.get_ascii_tree())

    deleted2 = bst2.delete_root_with_successor()
    print(f"\n删除的根节点值: {deleted2}")

    print("\n删除后 BST 形态：")
    print(bst2.get_ascii_tree())

    print("\n树的层级结构表示：")
    print("        60")
    print("       /  \\")
    print("      30   70")
    print("     / \\    \\")
    print("    20 40    80")

    print(f"\n中序遍历结果: {bst2.inorder_traversal()}")

    # ===== 回答混用问题 =====
    print("\n" + "=" * 70)
    print("【问题回答】两种删除节点的方法能混用吗？")
    print("=" * 70)
    print("""❌ 不能混用！

原因：
1. 中序前驱和中序后继是两种不同的删除策略，各自维护了BST的结构完整性
2. 如果在同一个删除操作中混用两种策略，会导致逻辑混乱和树结构破坏
3. 实际应用中，通常会固定选择一种策略（如始终使用后继），以保持代码一致性

重要区分：
- 中序前驱：左子树的最大值（最右节点）
- 中序后继：右子树的最小值（最左节点）

混合使用的后果：
- 可能违反BST性质（左<根<右）
- 可能导致节点重复或丢失
- 树的结构可能不再保持平衡或正确

正确做法：
- 在整个程序中统一使用一种删除策略
- 或者根据情况（如子树高度）动态选择，但不能在一次删除中同时使用两种
    """)

    print("\n" + "=" * 70)
    print("演示完成！")
    print("=" * 70)
```
