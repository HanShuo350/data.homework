## 第一步：整理图中的边

| 边 | 权重 |
|:---|:---:|
| A — B | 2 |
| B — C | 4 |
| A — D | 3 |
| B — E | 1 |
| C — F | 5 |
| D — E | 6 |
| E — F | 2 |

共 7 条边，6 个顶点，MST 需要包含 **5 条边**。

---

## 第二步：Prim 算法（从顶点 A 开始）

**思想**：从起点出发，每次选择连接已访问集合与未访问集合的最小权边。

| 步骤 | 已访问集合 | 可选边（跨集合） | 最小权边 | 加入 |
|:---:|:---|:---|:---:|:---|
| 1 | {A} | A—B(2), A—D(3) | **A—B(2)** | B |
| 2 | {A,B} | A—D(3), B—C(4), B—E(1) | **B—E(1)** | E |
| 3 | {A,B,E} | A—D(3), B—C(4), E—F(2), E—D(6) | **E—F(2)** | F |
| 4 | {A,B,E,F} | A—D(3), B—C(4), E—D(6) | **A—D(3)** | D |
| 5 | {A,B,D,E,F} | B—C(4), E—D(6) | **B—C(4)** | C |

### Prim 结果：
- **选中的边**：A—B(2), B—E(1), E—F(2), A—D(3), B—C(4)
- **总权重**：2 + 1 + 2 + 3 + 4 = **12**

---

## 第三步：Kruskal 算法

**思想**：按权重从小到大排序所有边，依次加入，不形成环则保留。

**边按权重排序**：

| 顺序 | 边 | 权重 | 是否加入 | 说明 |
|:---:|:---|:---:|:---:|:---|
| 1 | B—E | 1 | ✅ | 无环 |
| 2 | A—B | 2 | ✅ | 无环 |
| 3 | E—F | 2 | ✅ | 无环 |
| 4 | A—D | 3 | ✅ | 无环 |
| 5 | B—C | 4 | ✅ | 无环（此时已有5条边，完成）|
| 6 | C—F | 5 | ❌ | 会成环（C和F已连通）|
| 7 | D—E | 6 | ❌ | 会成环（D和E已连通）|

### Kruskal 结果：
- **选中的边**：B—E(1), A—B(2), E—F(2), A—D(3), B—C(4)
- **总权重**：1 + 2 + 2 + 3 + 4 = **12**

---

## 最终答案

| 算法 | 最小生成树的边 | 总权重 |
|:---|:---|:---:|
| **Prim** | A—B(2), B—E(1), E—F(2), A—D(3), B—C(4) | **12** |
| **Kruskal** | B—E(1), A—B(2), E—F(2), A—D(3), B—C(4) | **12** |

> 两种算法得到的 MST 边集相同（只是加入顺序不同），总权重均为 **12**，验证了结果的正确性。
